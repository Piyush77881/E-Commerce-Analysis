```python
import pandas as pd
import numpy as np
                                                    #pandas (pd): Data manipulation ke liye use hota hai, jaise CSV file load karna aur process karna.
import plotly.express as px                                                      #plotly.express (px): Data visualization library jo easy aur quick plots banata hai.
import plotly.graph_objects as go                                               # plotly.graph_objects (go): Advanced and customizable graphs banane ke liye
import plotly.io as pio                                                           #plotly.io (pio): Graph templates ko customize karne ke liye
import plotly.colors as colors
pio.templates.default = "plotly_white"
from plotly.subplots import make_subplots
```

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])
data['Ship Date'] = pd.to_datetime(data['Ship Date'])
```

```python
data.info()
```

```python
data['Order Month'] = data['Order Date'].dt.month
data['Order Year'] = data['Order Date'].dt.year
data['Order Day of Week'] = data['Order Date'].dt.dayofweek
```

```python
data.head()
```

```python
data.shape
```

```python
sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index()
```

```python
sales_by_month
```

```python

fig = px.line(sales_by_month,
              x='Order Month',
              y='Sales',
              title='Monthly Sales Analysis')
fig.show()
```

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()


fig = px.pie(sales_by_category,
             values='Sales',
             names='Category',
             hole=0.5,
             color_discrete_sequence=px.colors.qualitative.Pastel)

fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(title_text='Sales Analysis by Category', title_font=dict(size=24))

fig.show()
```

```python
sales_by_subcategory = data.groupby('Sub-Category')['Sales'].sum().reset_index()
fig = px.bar(sales_by_subcategory,
             x='Sub-Category',
             y='Sales',
             title='Sales Analysis by Sub-Category')
fig.show()
```

```python
profit_by_month = data.groupby('Order Month')['Profit'].sum().reset_index()
fig = px.line(profit_by_month,
              x='Order Month',
              y='Profit',
              title='Monthly Profit Analysis')
fig.show()
```

```python
profit_by_category = data.groupby('Category')['Profit'].sum().reset_index()

fig = px.pie(profit_by_category,
             values='Profit',
             names='Category',
             hole=0.5,
             color_discrete_sequence=px.colors.qualitative.Pastel)

fig.update_traces(textposition='inside', textinfo='percent+label')
fig.update_layout(title_text='Profit Analysis by Category', title_font=dict(size=24))

fig.show()
```

```python
profit_by_subcategory = data.groupby('Sub-Category')['Profit'].sum().reset_index()
fig = px.bar(profit_by_subcategory, x='Sub-Category',
             y='Profit',
             title='Profit Analysis by Sub-Category')
fig.show()
```

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()

color_palette = colors.qualitative.Pastel

fig = go.Figure()
fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'],
                     y=sales_profit_by_segment['Sales'],
                     name='Sales',
                     marker_color=color_palette[0]))

fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'],
                     y=sales_profit_by_segment['Profit'],
                     name='Profit',
                     marker_color=color_palette[1]))

fig.update_layout(title='Sales and Profit Analysis by Customer Segment',
                  xaxis_title='Customer Segment', yaxis_title='Amount')

fig.show()
```

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment', 'Sales_to_Profit_Ratio']])
```

```python

fig6 = px.scatter(
    data,
    x='Sales',
    y='Profit',
    color='Category',
    size='Quantity',
    hover_data=['City', 'Sub-Category'],
    title='Sales vs Profit Analysis (sized by Quantity)'
)

fig6.update_layout(height=600)
fig6.show()
```

```python

data = data.sample(min(500, len(data)))#3D Scatter - Sales, Profit, Quantity

fig7 = px.scatter_3d(
    data,
    x='Sales',
    y='Profit',
    z='Quantity',
    color='Category',
    size='Sales',
    hover_data=['City', 'Sub-Category'],
    title='3D Analysis: Sales, Profit & Quantity'
)

fig7.update_layout(height=700)
fig7.show()
```

```python

top_cities = data.groupby('City')['Sales'].sum().nlargest(15).index  #Heatmap - City vs Category

heatmap_data = data[data['City'].isin(top_cities)].pivot_table(
    values='Sales',
    index='City',
    columns='Category',
    aggfunc='sum',
    fill_value=0
)

fig8 = go.Figure(data=go.Heatmap(
    z=heatmap_data.values,
    x=heatmap_data.columns,
    y=heatmap_data.index,
    colorscale='Blues',
    text=heatmap_data.values,
    texttemplate='$%{text:,.0f}',
    textfont={"size": 9}
))

fig8.update_layout(
    title='Sales Heatmap: Top 15 Cities vs Category',
    xaxis_title='Category',
    yaxis_title='City',
    height=600
)

fig8.show()
```

```python
fig9 = px.box(                                                                            #Box Plot - Sales Distribution
    data,
    x='Category',
    y='Sales',
    color='Category',
    title='Sales Distribution by Category'
)

fig9.update_layout(height=500, showlegend=False)
fig9.show()
```

```python
fig10 = px.violin(  #Violin Plot - Profit Distribution
   data,
    x='Category',
    y='Profit',
    color='Category',
    box=True,
    title='Profit Distribution by Category'
)

fig10.update_layout(height=500, showlegend=False)
fig10.show()
```

```python
city_bubble = data.groupby('City').agg({                                                    #Bubble Chart - City Performance

    'Sales': 'sum',
    'Profit': 'sum',
    'Quantity': 'sum'
}).reset_index()

fig11 = px.scatter(
    city_bubble,
    x='Sales',
    y='Profit',
    size='Quantity',
    color='Profit',
    hover_name='City',
    title='City Performance: Sales vs Profit (sized by Quantity)',
    color_continuous_scale='RdYlGn'
)

fig11.update_layout(height=600)
fig11.show()
```

```python

top10_cities = data.groupby('City')['Sales'].sum().nlargest(10).reset_index()  #Funnel - Top 10 Cities

fig13 = go.Figure(go.Funnel(
    y=top10_cities['City'],
    x=top10_cities['Sales'],
    textinfo="value+percent initial"
))

fig13.update_layout(
    title='Top 10 Cities Sales Funnel',
    height=600
)

fig13.show()
```

```python
   #Parallel Categories
# ============================================================================

sample_parallel = data.sample(min(1000, len(data)))

fig14 = px.parallel_categories(
    sample_parallel,
    dimensions=['City', 'Category', 'Sub-Category'],
    color='Sales',
    color_continuous_scale='Turbo',
    title='Flow Analysis: City → Category → Sub-Category'
)

fig14.update_layout(height=600)
fig14.show()
```

```python
fig15 = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        'Sales by Category',
        'Profit by Category',
        'Top 10 Cities',
        'Quantity Distribution'
    ),
    specs=[[{'type': 'bar'}, {'type': 'bar'}],
           [{'type': 'bar'}, {'type': 'pie'}]]
)

# Sales by category
cat_sales = data.groupby('Category')['Sales'].sum().sort_values(ascending=False)
fig15.add_trace(
    go.Bar(x=cat_sales.index, y=cat_sales.values, marker_color='lightblue'),
    row=1, col=1
)

# Profit by category
cat_profit = data.groupby('Category')['Profit'].sum().sort_values(ascending=False)
fig15.add_trace(
    go.Bar(x=cat_profit.index, y=cat_profit.values, marker_color='lightgreen'),
    row=1, col=2
)

# Top 10 cities
top10 = data.groupby('City')['Sales'].sum().nlargest(10)
fig15.add_trace(
    go.Bar(x=top10.index, y=top10.values, marker_color='coral'),
    row=2, col=1
)

# Quantity pie
cat_qty = data.groupby('Category')['Quantity'].sum()
fig15.add_trace(
    go.Pie(labels=cat_qty.index, values=cat_qty.values),
    row=2, col=2
)

fig15.update_layout(
    height=900,
    showlegend=False,
    title_text='Comprehensive Dashboard'
)

fig15.show()
```
